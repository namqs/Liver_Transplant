```python
from pyspark.sql.functions import col, when, regexp_replace, round

# Substituir valores nulos por -2.0
df = df.withColumn(
    "BMI_DON_CALC",
    when(col("BMI_DON_CALC").isNull(), -2.0).otherwise(col("BMI_DON_CALC"))
)

# Remover os pontos (separadores de milhar) e converter para tipo numérico
df = df.withColumn(
    "BMI_DON_CALC",
    regexp_replace(col("BMI_DON_CALC"), r"\.", "")  # Remove os pontos
)

# Convertendo a coluna para tipo double (flutuante)
df = df.withColumn(
    "BMI_DON_CALC",
    col("BMI_DON_CALC").cast("double")
)

# Arredondar para 1 casa decimal, convertendo para string para evitar notação científica
df = df.withColumn(
    "BMI_DON_CALC",
    round(col("BMI_DON_CALC"), 1)
)

# Verificando os primeiros 20 valores após as transformações
df.select("BMI_DON_CALC").show(20, truncate=False)
```

```python

from pyspark.sql.functions import col, expr, when, round, coalesce
from pyspark.sql.types import FloatType, IntegerType
from pyspark.sql import functions as F

# 1. Transformar "PX_STAT" (A, D, R, L -> int)
df = df.withColumn("PX_STAT",
                   when(col("PX_STAT") == "A", 1)
                   .when(col("PX_STAT") == "D", 0)
                   .when(col("PX_STAT") == "R", 2)
                   .when(col("PX_STAT") == "L", 3)
                   .otherwise(-2))

# 2. Transformar "DIAB" (vazios -> -2, 998 -> -2)
df = df.withColumn("DIAB",
                   when(col("DIAB") == 998, -2)
                   .otherwise(when(col("DIAB").isNull(), -2).otherwise(col("DIAB"))))

# 3. Transformar "DIABETES_DON" (N -> 0, Y -> 1, A -> 2, vazios -> 0)
df = df.withColumn("DIABETES_DON",
                   when(col("DIABETES_DON") == "N", 0)
                   .when(col("DIABETES_DON") == "Y", 1)
                   .when(col("DIABETES_DON") == "A", -1)
                   .otherwise(-2))

# 4. Transformar "HIST_INSULIN_DEP_DON" (N -> 0, Y -> 1, A -> 2, vazios -> 0)
df = df.withColumn("HIST_INSULIN_DEP_DON",
                   when(col("HIST_INSULIN_DEP_DON") == "N", 0)
                   .when(col("HIST_INSULIN_DEP_DON") == "Y", 1)
                   .when(col("HIST_INSULIN_DEP_DON") == "A", -1)
                   .otherwise(-2))

# 5. Transformar "INIT_DIALYSIS_PRIOR_WEEK", "FINAL_DIALYSIS_PRIOR_WEEK" (N -> 0, Y -> 1, A -> 2, vazios -> 0)
for dialysis_col in ["INIT_DIALYSIS_PRIOR_WEEK", "FINAL_DIALYSIS_PRIOR_WEEK"]:
    df = df.withColumn(dialysis_col,
                       when(col(dialysis_col) == "N", 0)
                       .when(col(dialysis_col) == "Y", 1)
                       .when(col(dialysis_col) == "A", -1)
                       .otherwise(-2))

# 6. Transformar "GENDER", "GENDER_DON" (M -> 1, F -> 2)
for gender_col in ["GENDER", "GENDER_DON"]:
    df = df.withColumn(gender_col,
                       when(col(gender_col) == "M", 0)
                       .when(col(gender_col) == "F", 1)
                       .otherwise(-2))
# 7.
for col_name in ["DGN2_TCR", "INIT_MELD_PELD_LAB_SCORE"]:
    df = df.withColumn(
        col_name,
        when(col(col_name).isNull(), -2)  # Substituir NULL por -2
        .when(col(col_name) == 999, -2)  # Substituir 999 por -2
        .otherwise(col(col_name))       # Manter os valores restantes
    )

# 8. Alterar valores vazios para "INIT_SERUM_SODIUM", "FINAL_SERUM_SODIUM" para 0
for sodium_col in ["INIT_SERUM_SODIUM", "FINAL_SERUM_SODIUM"]:
    df = df.withColumn(sodium_col,
                       when(col(sodium_col).isNull(), -2).otherwise(col(sodium_col)))

# 9. Transformar "PREV_TX", "PREV_TX_ANY" (N -> 0, Y -> 1, A -> 2, vazios -> 0)
for tx_col in ["PREV_TX", "PREV_TX_ANY"]:
    df = df.withColumn(tx_col,
                       when(col(tx_col) == "N", 0)
                       .when(col(tx_col) == "Y", 1)
                       .when(col(tx_col) == "A", -1)
                       .otherwise(-2))

# 10. Alterar valores vazios e 999 para "COD_CAD_DON" para 0
df = df.withColumn("COD_CAD_DON",
                   when((col("COD_CAD_DON") == 999) | (col("COD_CAD_DON").isNull()), -2)
                   .otherwise(col("COD_CAD_DON")))

# 11. Alterar valores vazios para "COLD_ISCH" para 0.0
df = df.withColumn(
    "COLD_ISCH",
    round(
        when(col("COLD_ISCH").isNull(), -2.0)
        .otherwise(col("COLD_ISCH").cast(FloatType())), 1
    )
)
# 12.

# 13. Transformar "DATA_WAITLIST" (N -> 0, Y -> 1, A -> -1, vazios -> 0)
df = df.withColumn("DATA_WAITLIST",
                   when(col("DATA_WAITLIST") == "N", 0)
                   .when(col("DATA_WAITLIST") == "Y", 1)
                   .when(col("DATA_WAITLIST") == "A", -1)
                   .otherwise(-2))
# 14.
df = df.withColumn(
    "CREAT_DON",
    when(col("CREAT_DON").isNull(), -2.0).otherwise(col("CREAT_DON"))
)

# Mostrar a tabela transformada
df.show()
```

```python
# Instalar bibliotecas necessárias
!pip install pyspark gdown

# Importar as bibliotecas necessárias
import gdown
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, udf, when
from pyspark.sql.types import DoubleType

# Criar sessão do Spark
spark = SparkSession.builder.appName("Processar BMI_DON_CALC").getOrCreate()

# Função para processar a coluna
def process_bmi(value):
    if value is None:
        return None
    # Remover pontos
    cleaned_value = value.replace(".", "")
    # Pegar os 3 primeiros caracteres
    first_three_digits = cleaned_value[:3]
    # Converter para double com formato de ponto flutuante "00.0"
    try:
        return float(first_three_digits[:2] + "." + first_three_digits[2])
    except ValueError:
        return None

# Registrar UDF
process_bmi_udf = udf(process_bmi, DoubleType())

# URL do arquivo no Google Drive
file_url = "https://drive.google.com/uc?id=1-3MihaLxViOob-1E23pxj_8aYVzBtw9m"  # Substitua pelo link de compartilhamento
output_file = "/content/tabela_original.csv"  # Caminho para salvar o arquivo baixado
gdown.download(file_url, output_file, quiet=False)

# Carregar a tabela no Spark
input_path = "/content/tabela_original.csv"
df = spark.read.csv(input_path, header=True, inferSchema=True)

# Aplicar a função de processamento na coluna BMI_DON_CALC
df = df.withColumn("BMI_DON_CALC", process_bmi_udf(col("BMI_DON_CALC")))

# Substituir valores nulos com um valor padrão (opcional)
df = df.fillna({"BMI_DON_CALC": -2.0})

# Salvar a nova tabela como um único arquivo CSV no Google Drive
output_path = "/content/tabela_modificada.csv"
df.coalesce(1).write.csv(output_path, header=True, mode="overwrite")

# Fazer upload do arquivo para o Google Drive
from google.colab import drive
import shutil

drive.mount('/content/drive')
shutil.move(output_path, '/content/drive/My Drive/tabela_modificada.csv')

print("Tabela processada e salva no Google Drive com sucesso!")
```
